```python
%cd /content/ComfyUI

!pip install -q pinggy

import subprocess
import threading
import time
from pinggy import start_tunnel

# --------------------------------------------------
# 1. Kill any old ComfyUI / tunnel processes
# --------------------------------------------------

!pkill -f "python main.py" || true
!pkill -f cloudflared || true

# --------------------------------------------------
# 2. Start ComfyUI in the background
# --------------------------------------------------

print("Starting ComfyUI...")

comfy = subprocess.Popen(
    [
        "python",
        "main.py",
        "--listen", "0.0.0.0",
        "--port", "8188"
    ],
    stdout=open("/content/comfyui.log", "w"),
    stderr=subprocess.STDOUT
)

# Give ComfyUI time to initialize
time.sleep(15)

# --------------------------------------------------
# 3. Check ComfyUI
# --------------------------------------------------

import requests

try:
    r = requests.get(
        "http://127.0.0.1:8188/system_stats",
        timeout=5
    )

    if r.status_code == 200:
        print("✅ ComfyUI is running")
    else:
        print("⚠️ ComfyUI returned:", r.status_code)

except Exception as e:
    print("❌ ComfyUI isn't responding:", e)
    print("\n--- LOG ---")
    print(open("/content/comfyui.log").read())
    raise

# --------------------------------------------------
# 4. Start Pinggy tunnel
# --------------------------------------------------

print("\n🌐 Starting Pinggy tunnel...")
print("The public HTTPS URL will appear below.\n")

def run_tunnel():
    start_tunnel(port=8188)

threading.Thread(
    target=run_tunnel,
    daemon=True
).start()

# Keep notebook alive
while True:
    time.sleep(60)
```

```python
%cd /content/ComfyUI

import subprocess
import time
import requests

# Start ComfyUI
comfy = subprocess.Popen(
    [
        "python", "main.py",
        "--listen", "0.0.0.0",
        "--port", "8188",
        "--enable-cors-header", "*"
    ],
    stdout=subprocess.PIPE,
    stderr=subprocess.STDOUT,
    text=True
)

# Wait until ComfyUI is actually ready
for i in range(60):
    try:
        r = requests.get("http://127.0.0.1:8188/system_stats", timeout=2)
        if r.status_code == 200:
            print("✅ ComfyUI ready")
            break
    except:
        time.sleep(2)
else:
    raise RuntimeError("ComfyUI failed to start")
```
